```python
from secrets import compare_digest
from pqcrypto.kem.mceliece8192128 import generate_keypair, encrypt, decrypt

# Message we want to encrypt
shared_secret = b"Hello World!"

# Step 1: Generate post-quantum key pair
public_key, secret_key = generate_keypair()

# Step 2: Encrypt (encapsulate) a shared secret
ciphertext, original_secret = encrypt(public_key)

# Double-check they match
assert compare_digest(shared_secret, original_secret) == False, \
       "note: KEM generates its own secret—this one is random"

# Step 3: Decrypt the ciphertext to recover the shared secret
recovered_secret = decrypt(secret_key, ciphertext)
assert compare_digest(recovered_secret, original_secret)

print("Ciphertext (hex):", ciphertext.hex())
print("Recovered shared secret (hex):", recovered_secret.hex())
```
